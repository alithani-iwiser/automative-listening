```python
def clean_text(x):
    if pd.isna(x):
        return ""
    x = html.unescape(str(x))
    x = re.sub(r"https?://\\S+", " ", x)
    x = re.sub(r"\\s+", " ", x)
    return x.strip().lower()

df["content_clean"] = df["content"].map(clean_text)
df["topics_clean"] = df["topics"].map(clean_text)
```

# Cell 7 — Apply the Step 1 weak filter

```python
# ============================================================
# CELL 7 — APPLY STEP 1 WEAK FILTER
# ============================================================

# Content is the main signal.
# Topics are a supporting signal to maximize recall.
# Domain is NOT used to decide automotive relevance.
#
# Step 1 is intentionally WEAK:
# We want to keep anything that MIGHT possibly be related
# to automotive, even if it turns out to be irrelevant later.

df["url_clean"] = df["url"].map(clean_text)

df["scan_text"] = (
    df["content_clean"].fillna("") + " " +
    df["topics_clean"].fillna("")
).str.strip()

df["hit_car_term"] = df["scan_text"].map(
    lambda x: bool(CAR_RX.search(x) or CJK_RX.search(x))
)

df["hit_url_auto_term"] = df["url_clean"].map(
    lambda x: bool(URL_AUTO_RX.search(x))
)

df["hit_context_term"] = df["scan_text"].map(
    lambda x: bool(CONTEXT_RX.search(x))
)

df["step1_keep"] = (
    df["hit_car_term"] |
    df["hit_url_auto_term"] |
    df["hit_context_term"]
)
```

```python
# ============================================================
# CELL 8 — STEP 1 FILTER SUMMARY
# ============================================================

total = len(df)
kept = int(df["step1_keep"].sum())
removed = total - kept

car_only = int(
    (df["hit_car_term"] & ~df["hit_context_term"]).sum()
)

context_only = int(
    (~df["hit_car_term"] & df["hit_context_term"]).sum()
)

both = int(
    (df["hit_car_term"] & df["hit_context_term"]).sum()
)

neither = int(
    (~df["hit_car_term"] & ~df["hit_context_term"]).sum()
)

print("=" * 60)
print("STEP 1 WEAK FILTER SUMMARY")
print("=" * 60)

print(f"Total rows          : {total:,}")
print(f"Step 1 kept         : {kept:,} ({kept/total:.1%})")
print(f"Step 1 removed      : {removed:,} ({removed/total:.1%})")

print()
print("WHY ROWS WERE KEPT")
print("-" * 60)
print(f"Car/brand/model only: {car_only:,}")
print(f"Context terms only  : {context_only:,}")
print(f"Both                : {both:,}")
print(f"Neither             : {neither:,}")

print()
print("CHECK")
print("-" * 60)
print(f"Kept + Removed      : {kept + removed:,}")
print(f"Original total      : {total:,}")

if kept + removed == total:
    print("✓ Row count is consistent.")
else:
    print("⚠ Row count mismatch — investigate.")
```

```python
# ============================================================
# CELL 10 — SAVE STEP 1 OUTPUT + REMOVED ROWS
# ============================================================

# Read the additional columns from the original Excel file
extra = pd.read_excel(
    FILE_PATH,
    usecols="H,J,AW:AZ,BB"
)

print("Original headers found:")
print(extra.columns.tolist())

# Rename them to consistent names for our output
extra.columns = [
    "date (UTC)",
    "language",
    "no_of_comments",
    "no_of_likes",
    "no_of_shares",
    "no_of_views",
    "engagement"
]

# Make sure the rows still line up with the original dataframe
assert len(extra) == len(df), (
    f"ERROR: Row count mismatch. "
    f"df = {len(df):,}, extra = {len(extra):,}"
)

# Add the columns to our existing dataframe
for col in extra.columns:
    df[col] = extra[col].values

print()
print("Additional columns successfully added:")
print(extra.columns.tolist())


# ------------------------------------------------------------
# Select the columns we want in the final workbook
# ------------------------------------------------------------

output_columns = [
    "source",
    "content",
    "url",
    "domain",
    "topics",
    "sentiment",
    "sentiment class",
    "date (UTC)",
    "language",
    "no_of_comments",
    "no_of_likes",
    "no_of_shares",
    "no_of_views",
    "engagement",
    "hit_car_term",
    "hit_context_term",
    "step1_keep"
]


# ------------------------------------------------------------
# Separate Step 1 kept and removed rows
# ------------------------------------------------------------

step1_df = df.loc[
    df["step1_keep"],
    output_columns
].copy()

step1_removed_df = df.loc[
    ~df["step1_keep"],
    output_columns
].copy()


# ------------------------------------------------------------
# Save both into the SAME Excel workbook
# ------------------------------------------------------------

with pd.ExcelWriter(
    OUTPUT_PATH,
    engine="openpyxl"
) as writer:

    step1_df.to_excel(
        writer,
        sheet_name="Step1_Kept",
        index=False
    )

    step1_removed_df.to_excel(
        writer,
        sheet_name="Step1_Removed",
        index=False
    )


# ------------------------------------------------------------
# Final verification
# ------------------------------------------------------------

print()
print("=" * 70)
print("STEP 1 WORKBOOK CREATED")
print("=" * 70)

print(f"File: {OUTPUT_PATH}")
print(f"Step 1 kept    : {len(step1_df):,}")
print(f"Step 1 removed : {len(step1_removed_df):,}")
print(f"Total          : {len(step1_df) + len(step1_removed_df):,}")
print(f"Original df    : {len(df):,}")

assert (
    len(step1_df) + len(step1_removed_df)
    == len(df)
)

print()
print("Sheets:")
print("1. Step1_Kept")
print("2. Step1_Removed")
print()
print("✓ Complete.")
```
